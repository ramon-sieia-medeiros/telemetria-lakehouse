```python
%sql
-- 1. Volume e arquivos
SELECT COUNT(*) AS linhas, COUNT(DISTINCT _source_file) AS arquivos
FROM project_lakehouse_telemetria_ele.bronze.telemetria_raw;

-- 2. Nulos por coluna
SELECT SUM(CASE WHEN tensao_v IS NULL THEN 1 ELSE 0 END) AS tensao_nula,
       SUM(CASE WHEN corrente_a = '-999' THEN 1 ELSE 0 END) AS corrente_sentinela
FROM project_lakehouse_telemetria_ele.bronze.telemetria_raw;

-- 3. Duplicados
SELECT COUNT(*) - COUNT(DISTINCT device_id, event_ts) AS duplicados
FROM project_lakehouse_telemetria_ele.bronze.telemetria_raw;

-- 4. Leituras por dispositivo
SELECT device_id, COUNT(*) AS leituras
FROM project_lakehouse_telemetria_ele.bronze.telemetria_raw
GROUP BY device_id ORDER BY leituras DESC LIMIT 10;

-- 5. Temperatura média por hora do dia (curva de carga)
SELECT hour(CAST(event_ts AS TIMESTAMP)) AS hora,
       ROUND(AVG(CAST(temperatura_c AS DOUBLE)), 1) AS temp_media
FROM project_lakehouse_telemetria_ele.bronze.telemetria_raw
GROUP BY 1 ORDER BY 1;
```
